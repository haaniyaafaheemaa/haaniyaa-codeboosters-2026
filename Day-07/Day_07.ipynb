```python
import pandas as pd
import numpy as np

from sentence_transformers import SentenceTransformer
import chromadb
print('All Libraries imported successfully!')
print(f"ChromaDB version: {chromadb.__version__}")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully")
print(f"Model produces vectors of size:{model.get_sentence_embedding_dimension()} dimensions")
```

```python
sentence = "ETL is used to clean and transform the data"
embedding = model.encode(sentence)
print(f"Input sentence:'{sentence}'")

print(f"Embedding type: {type(embedding)}")

print(f"Embedding shape:{embedding.shape}")

print(f"First 10 numbers:{embedding[:10].round(4)}")

print(f"Min value:{embedding.min():.4f}")

print(f"Max vslue: {embedding.max():.4f}")
```

```python
documents=["Cars and trucks are popular automobiles",
          "SQL is used to query databases",
           "Machine Learning trains models on data"
]

query_keyboard="vehicle"

print(f"beyond search for:'{query_keyboard}")

for i, doc in enumerate(documents):
  if query_keyboard.lower() in doc.lower():
    print(f" Found [doc_{i}]:{doc}")
  else:
    print(f"Missed[doc_{i}]:{doc}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
documents = [
    "Cars and trucks are popular automobiles",
    "SQL is used to query databases",
    "Machine Learning trains models on data"
]

query = "vehicle"

print(f"Semantic search for: '{query}'\n")

doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])


scores = cosine_similarity(query_embedding, doc_embeddings)[0]

threshold = 0.5

for i, score in enumerate(scores):
    if score >= threshold:
        print(f"Found [doc_{i}] (score={score:.2f}): {documents[i]}")
    else:
        print(f"Missed [doc_{i}] (score={score:.2f}): {documents[i]}")

best_match = scores.argmax()
print("Best Match:")
print(documents[best_match])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

your_sentences = [
    "Switzerland is one of the best countries to explore",
    "I love eating healthy foods",
    "AI algorithms learn pattern from data"
]

your_embeddings = model.encode(your_sentences)

sim_your_01 = cosine_similarity(
    [your_embeddings[0]],
    [your_embeddings[1]]
)

sim_your_02 = cosine_similarity(
    [your_embeddings[0]],
    [your_embeddings[2]]
)

print('YOUR EXPERIMENT RESULTS')
print(f"Sentence A: '{your_sentences[0]}'")
print(f"Sentence B: '{your_sentences[1]}'")
print(f"Sentence C: '{your_sentences[2]}'")
print()

print(f"Similarity(A vs B): {sim_your_01[0][0]:.4f}")
print(f"Similarity(A vs C): {sim_your_02[0][0]:.4f}")
```

```python
chroma_client = chromadb.Client()
collection = chroma_client.get_or_create_collection('demo_notes')

print("Documents in collection:")
print(collection.count())
```

```python
sample_docs = [
    "ETL means Extract, Transform and Load",
    "SQL is used to query databases",
    "Machine Learning trains models on data",
    "Python Pandas library is used for data manipulation and cleaning",
    "Neural networks are inspired by how the human brain works"
]

sample_ids = ["doc_1", "doc_2", "doc_3", "doc_4", "doc_5"]
sample_metadata = [
    {"subject": "Data Engineering", "topic":"ETL"},
    {"subject": "Data Engineering", "topic":"SQL"},
    {"subject": "Machine Learning", "topic":"ML Basics"},
    {"subject": "Python", "topic":"Pandas"},
    {"subject": "Machine Learning", "topic":"Neural Networks"}

]

collection.add(
    documents=sample_docs,
    ids=sample_ids,
    metadatas=sample_metadata
)

print("Documents in collection:")
print(collection.count())
```

```python
collection.delete(ids=["doc_3"])
print("Documents in collection after deleting:")
print(collection.count())
```

```python
query = "database"

results = collection.query(
    query_texts=[query],
    n_results=3
)
print(f"Query: '{query}'")
print()

matched_docs = results['documents'][0]
matched_ids = results['ids'][0]
matched_distances = results['distances'][0]
matched_metadata = results['metadatas'][0]

for rank, (doc, doc_id, dist, meta) in enumerate(
    zip(matched_docs, matched_ids, matched_distances, matched_metadata),
    start=1
):
    print(f"Rank: {rank}")
    print(f"ID: {doc_id}")
    print(f"Distance: {dist:.4f}")
    print(f"Subject: {meta['subject']} | Topic: {meta['topic']}")
    print(f"Document: {doc}")
    print()
```
